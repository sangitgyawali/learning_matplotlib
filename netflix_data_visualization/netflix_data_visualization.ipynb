```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("netflix_titles.csv")

type_counts = df['type'].value_counts()

plt.bar(type_counts.index, type_counts.values, color=['skyblue', 'coral'])
plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Type")
plt.ylabel("Count")
plt.show()

year_counts = df['release_year'].value_counts().sort_index()

plt.plot(year_counts.index, year_counts.values, marker='o', color='green')
plt.title("Netflix Releases per Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.show()

country_counts = df['country'].value_counts().head(10)

plt.barh(country_counts.index, country_counts.values, color='orange')
plt.title("Top 10 Countries with Most Netflix Content")
plt.xlabel("Number of Titles")
plt.ylabel("Country")
plt.show()

rating_counts = df['rating'].value_counts().head(6)

plt.pie(rating_counts.values, labels=rating_counts.index, autopct='%1.1f%%',
        colors=['gold', 'skyblue', 'lightgreen', 'coral', 'violet', 'pink'])
plt.title("Distribution of Ratings on Netflix")
plt.show()
```
