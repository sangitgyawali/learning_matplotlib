```python
import matplotlib.pyplot as plt

regions = ['North', 'South', 'East', 'West']
revenue = [3000, 2000, 1500, 1000]
x = [1, 2, 3, 4]
fig, axs = plt.subplots(1, 2, figsize=(10, 5))

# Chart 1: Bar chart
axs[0].bar(regions, revenue, color='skyblue')
axs[0].set_title('Revenue by Region (Bar Chart)')
axs[0].set_xlabel('Regions')
axs[0].set_ylabel('Revenue')

# Chart 2: Scatter plot
axs[1].scatter(x, revenue, color='coral', s=100)
axs[1].set_title('Revenue by Region (Scatter Plot)')
axs[1].set_xlabel('Regions')
axs[1].set_ylabel('Revenue')
axs[1].set_xticks(x)
axs[1].set_xticklabels(regions)

plt.tight_layout()
plt.show()
```
