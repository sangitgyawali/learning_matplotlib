```python
import matplotlib.pyplot as plt

months = [1, 2, 3, 4]
sales = [1000, 1500, 1200, 1800]

plt.plot(months, sales, color='blue', linestyle='--', linewidth=2, marker='o', label='2025 sales data')
plt.xlabel('Months')
plt.ylabel('Sales per month')
plt.title('Monthly Sales Data Report')
plt.legend(loc='upper left', fontsize=12)
plt.grid(color='gray', linestyle=':', linewidth=1)
plt.xlim(1, 4)
plt.ylim(0, 2000)
plt.xticks([1, 2, 3, 4], ['M1', 'M2', 'M3', 'M4'])
plt.show()
```

```python
# Plotting Bar graph in matplotlib
import matplotlib.pyplot as plt

product = ['A', 'B', 'C', 'D']
sales = [1000, 1500, 800, 1200]

plt.bar(product, sales, color='orange', label='Sales in 2025')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.title('Product Sale Comparision')
plt.legend()
plt.show()
```

```python
# Plotting Pie chart in matplotlib
import matplotlib.pyplot as plt

regions = ['North', 'South', 'East', 'West']
revenue = [3000, 2000, 1500, 1000]

plt.pie(revenue, labels=regions, autopct='%1.1f%%', colors=['gold', 'skyblue', 'lightgreen', 'coral'])
plt.xlabel('Products')
plt.ylabel('Sales')
plt.title('Product Sale Comparision')
plt.legend()
plt.show()
```

```python
# Plotting histogram
import matplotlib.pyplot as plt

sales = [3000, 2000, 1500, 1000, 1800, 2200, 2700, 3500, 1200, 1700]

plt.hist(sales, bins=5, color='skyblue', edgecolor='black')
plt.xlabel('Sales Range')
plt.ylabel('Frequency')
plt.title('Sales Distribution (Histogram)')
plt.legend(['Sales'])
plt.show()
```

```python
import matplotlib.pyplot as plt

regions = ['North', 'South', 'East', 'West']
revenue = [3000, 2000, 1500, 1000]

x = [1, 2, 3, 4]

plt.scatter(x, revenue, color='coral', marker='o', s=100)
plt.xlabel('Regions')
plt.ylabel('Revenue')
plt.title('Region-wise Revenue (Scatter Plot)')
plt.xticks(x, regions)
plt.legend(['Revenue'])
plt.show()
```
